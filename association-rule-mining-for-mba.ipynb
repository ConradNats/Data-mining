  <h1 style="margin: 0; color: darkblue;">Apriori Algorithm for MBA</h1>
</div>

The Apriori algorithm is a foundational data mining technique used to identify frequent itemsets and generate association rules in a dataset. It operates on the principle that all non-empty subsets of a frequent itemset must also be frequent. Here's an in-depth explanation:

### **<span style="background-color: darkblue; padding: 3px; border-radius: 3px;">How the Apriori Algorithm Works</span>**

1. **Generate Candidate Itemsets:**
   The algorithm starts by generating candidate itemsets of size 1. It then scans the dataset to count the support of each itemset and removes those that do not meet the minimum support threshold.

2. **Generate Larger Itemsets:**
   Using the frequent itemsets of size 1, the algorithm generates candidate itemsets of size 2, then size 3, and so on. For each size, it scans the dataset to count support and filters out infrequent itemsets.

3. **Generate Association Rules:**
   Once all frequent itemsets are identified, the algorithm generates association rules from these itemsets. Each rule is evaluated based on its confidence and lift, and only those that meet the minimum thresholds are retained.

### **<span style="background-color: darkblue; padding: 3px; border-radius: 3px;">Advantages of the Apriori Algorithm</span>**

- **Simplicity:** Easy to understand and implement.
- **Comprehensive:** Can handle large datasets and find all possible frequent itemsets.

### **<span style="background-color: darkblue; padding: 3px; border-radius: 3px;">Limitations</span>**

- **Scalability:** Can be computationally expensive for very large datasets.
- **Redundancy:** May generate a large number of candidate itemsets, many of which may not be useful.

### **<span style="background-color: darkblue; padding: 3px; border-radius: 3px;">Key Metrics</span>**

We're focusing on the key metrics because they have the most impact on the results:

1. **Support:**
   - **Definition:** Support measures how frequently an itemset appears in the dataset.
   - **Equation:**
     $$
     \text{Support}(A) = \frac{\text{Number of transactions containing item A}}{\text{Total number of transactions}}
     $$
   - **Example:** If item A appears in 30 out of 100 transactions, the support is 0.3 or 30%.

2. **Confidence:**
   - **Definition:** Confidence measures the likelihood that an item B is purchased when item A is purchased.
   - **Equation:**
     $$
     \text{Confidence}(A \rightarrow B) = \frac{\text{Support}(A \cup B)}{\text{Support}(A)}
     $$
   - **Example:** If 20 out of the 30 transactions containing item A also contain item B, the confidence is 0.67 or 67%.

3. **Lift:**
   - **Definition:** Lift measures how much more likely items A and B are to be purchased together than expected if they were independent.
   - **Equation:**
     $$
     \text{Lift}(A \rightarrow B) = \frac{\text{Confidence}(A \rightarrow B)}{\text{Support}(B)}
     $$
   - **Example:** If the support of item B is 0.2 and the confidence is 0.67, the lift is 3.35.

### **<span style="background-color: darkblue; padding: 3px; border-radius: 3px;">Supporting Metrics</span>**

1. **Leverage:**
   - **Definition:** Leverage measures the difference between the observed frequency of the itemset and the expected frequency if the items were independent.
   - **Equation:**
     $$
     \text{Leverage}(A \rightarrow B) = \text{Support}(A \cup B) - (\text{Support}(A) \times \text{Support}(B))
     $$
   - **Impact:** Higher leverage indicates a stronger association, useful for guiding marketing strategies.

2. **Conviction:**
   - **Definition:** Conviction measures the ratio of the expected frequency of the antecedent occurring without the consequent to the observed frequency of the antecedent occurring with the consequent.
   - **Equation:**
     $$
     \text{Conviction}(A \rightarrow B) = \frac{1 - \text{Support}(B)}{1 - \text{Confidence}(A \rightarrow B)}
     $$
   - **Impact:** Higher conviction values indicate stronger rules, useful for confident decision-making in promotions.

3. **Zhang's Metric:**
   - **Definition:** Zhang's metric measures the strength of the rule based on confidence and unexpectedness.
   - **Equation:**
     $$
     \text{Zhang's Metric}(A \rightarrow B) = \frac{\text{Confidence}(A \rightarrow B) - \text{Support}(B)}{1 - \text{Support}(B)}
     $$
   - **Impact:** A high Zhang’s Metric indicates a strong and unexpected rule, valuable for uncovering hidden patterns.

4. **Jaccard:**
   - **Definition:** Jaccard similarity measures the overlap between the antecedent and consequent.
   - **Equation:**
     $$
     \text{Jaccard}(A \rightarrow B) = \frac{\text{Support}(A \cup B)}{\text{Support}(A) + \text{Support}(B) - \text{Support}(A \cup B)}
     $$
   - **Impact:** Higher Jaccard similarity suggests a stronger overlap, useful for decisions on product placement and cross-promotion.

5. **Certainty Factor:**
   - **Definition:** Certainty factor quantifies how much more likely the consequent is given the antecedent compared to its baseline likelihood.
   - **Equation:**
     $$
     \text{Certainty Factor}(A \rightarrow B) = \frac{\text{Confidence}(A \rightarrow B) - \text{Support}(B)}{1 - \text{Support}(B)}
     $$
   - **Impact:** High certainty factors suggest a robust relationship, helpful in deciding on confident marketing actions.

6. **Kulczynski:**
   - **Definition:** Kulczynski measures the average of the two confidence values (antecedent to consequent and consequent to antecedent).
   - **Equation:**
     $$
     \text{Kulczynski}(A \rightarrow B) = \frac{1}{2}\left(\text{Confidence}(A \rightarrow B) + \text{Confidence}(B \rightarrow A)\right)

Set a minimum support threshold of 3 transactions (20%).

### **Step 2**: Generate Candidate Itemsets of Size 2

Generate triplets of items and count their support:

### **Step 4**: Generate Association Rules

Generate association rules from the frequent itemsets of size 3:

   - **Support:** 3/15 = 0.20

### **Step 5**: Interpreting the Results

#### (Milk, Bread) -> Butter

- **Support (0.20):** This means that 20% of all transactions contain the combination of Milk, Bread, and Butter.
- **Confidence (0.60):** Indicates that 60% of the transactions that include Milk and Bread also include Butter. This suggests a strong association between these items.
- **Lift (1.00):** A lift of 1.00 means that the occurrence of Butter is exactly what would be expected if Milk and Bread were independent of Butter. This indicates that there is no strong correlation between buying Butter and the combination of Milk and Bread beyond what would be expected by chance.

#### (Bread, Butter) -> Orange Juice

- **Support (0.20):** This means that 20% of all transactions contain the combination of Bread, Butter, and Orange Juice.
- **Confidence (0.50):** Indicates that 50% of the transactions that include Bread and Butter also include Orange Juice. This suggests a moderate association between these items.
- **Lift (1.25):** A lift of 1.25 means that Orange Juice is 25% more likely to be bought with Bread and Butter compared to it being bought independently. This indicates a positive correlation between buying Orange Juice and the combination of Bread and Butter.

For educational purposes, will use the [Market Basket Analysis Data](https://www.kaggle.com/datasets/ahmtcnbs/datasets-for-appiori). by [@AHMET BAŞ](https://www.kaggle.com/ahmtcnbs)

```python
import pandas as pd
Dataset = pd.read_csv('/kaggle/input/datasets-for-appiori/basket_analysis.csv', index_col=0)
Dataset
```

The dataset is allready in one-hot-encoded and ready to use by the algorithm, but in real word scinario, data preprocessing and transformation is neccesary

### **<span style="background-color: lightblue; padding: 3px; border-radius: 3px;">Apply the Apriori Algorithm</span>**
We will use the `apriori` function from the `mlxtend` library to find frequent itemsets and generate association rules.

Our dataset consists of 999 transactions. For a dataset of this size, setting too high a min_support threshold might exclude interesting patterns, while too low a threshold could result in too many itemsets, making the analysis computationally intensive.

We aim to capture a meaningful number of itemsets while maintaining computational efficiency. A min_support threshold of 0.02 means that an itemset must appear in at least 2% of the transactions to be considered frequent. For our dataset, this translates to approximately 20 transactions.

so setting min_support to 0.02 strikes a balance between capturing both frequent and slightly less frequent but potentially interesting patterns. It ensures that we do not miss out on significant itemsets that could provide valuable insights for decision-making.

```python
from mlxtend.frequent_patterns import apriori, association_rules

# Apply the apriori algorithm to find frequent itemsets
frequent_itemsets = apriori(Dataset, min_support=0.02, use_colnames=True)

# Display the frequent itemsets
print(frequent_itemsets)
```

### **<span style="background-color: darkblue; padding: 3px; border-radius: 3px;">Generate Association Rules</span>**
Now we can generate association rules from teh frequent itemsets.

```python
# Generate association rules from the frequent itemsets 
rules = association_rules(frequent_itemsets, metric="lift", num_itemsets=len(frequent_itemsets), min_threshold=1) 
# Calculate itemset size as total number of items in both antecedents and consequents
rules['itemset_size'] = rules['antecedents'].apply(len) + rules['consequents'].apply(len)

# Combine antecedents and consequents into a single string for labels
rules['itemset'] = rules['antecedents'].apply(lambda x: ', '.join(list(x))) + ' -> ' + rules['consequents'].apply(lambda x: ', '.join(list(x)))

# Display the association rules 
rules.sort_values(by="lift", ascending=False).head(3)
```

### **<span style="background-color: darkblue; padding: 3px; border-radius: 3px;">Visualize The Results</span>** 
Now after we get all the associations with their results, we can visualize the key metrics to better understand and interpret the data. Visualization helps in identifying patterns, trends, and insights that might not be immediately apparent from the raw data. Here are some visualizations we can create:

```python
### Support vs Lift by Itemset Size
import matplotlib.pyplot as plt
import seaborn as sns

# Create a scatter plot
plt.figure(figsize=(12, 8))
scatter = plt.scatter(
    rules['support'], 
    rules['lift'], 
    c=rules['itemset_size'],  # Color based on itemset size
    cmap='viridis',  # Color map
    alpha=0.7,  # Slight transparency
    s=100  # Marker size
)

# Add colorbar with intervals of 1 
cbar = plt.colorbar(scatter, label='Itemset Size') 
cbar.set_ticks(list(range(2, rules['itemset_size'].max() + 1)))

# Labels and title
plt.xlabel('Support')
plt.ylabel('Lift')
plt.title('Association Rules: Support vs Lift, Colored by Itemset Size')

# Add grid for readability
plt.grid(True, linestyle='--', alpha=0.7)

# Show the plot
plt.tight_layout()
plt.show()
```

```python
### Average support and Lift by Itemset Size

# Group by itemset size and calculate average support and lift
average_metrics = rules.groupby('itemset_size').agg({'support': 'mean', 'lift': 'mean'}).reset_index()

fig, ax1 = plt.subplots(figsize=(12, 6))

# Plot average lift on the primary y-axis
ax1.plot(average_metrics['itemset_size'], average_metrics['lift'], marker='o', color='b', label='Average Lift')
ax1.set_xlabel('Itemset Size')
ax1.set_ylabel('Average Lift', color='b')
ax1.tick_params(axis='y', labelcolor='b')

# Create a secondary y-axis to plot average support
ax2 = ax1.twinx()
ax2.plot(average_metrics['itemset_size'], average_metrics['support'], marker='o', color='g', label='Average Support')
ax2.set_ylabel('Average Support', color='g')
ax2.tick_params(axis='y', labelcolor='g')

# Adding title and legend
plt.title('Average Support and Lift by Itemset Size')
fig.tight_layout()

# Show the plot
plt.show()
```

It's clearly appears that on average, as the itemset size increases:

- **Support decreases**: Larger itemsets (containing more items) are less frequent, as it's rarer for multiple specific items to be purchased together.

- **Lift increases**: Larger itemsets often have higher lift, indicating stronger associations among the items within those sets when they do occur.

Let's filtered the rules dataset to show the top point with the highest lift and support for each itemset size (cluster). This way, the business Decission will be more specific.

```python
### Top itemset by Lift and Support

# Sort by itemset size, then by lift and support
rules_sorted = rules.sort_values(by=['itemset_size', 'lift', 'support'], ascending=[True, False, False])

# Group by itemset size and select the top itemset in each group
top_rules = rules_sorted.groupby('itemset_size').head(1).reset_index(drop=True)

# Display the filtered rules
top_rules
```

Let's also visualize how the average confidence of association rules varies with the size of the itemsets. 

```python
### Average Confidence by Itemset Size

# Group by itemset size and calculate average confidence
avg_confidence = rules.groupby('itemset_size')['confidence'].mean().reset_index()

# Sort the DataFrame by itemset size in descending order
avg_confidence = avg_confidence.sort_values(by='itemset_size', ascending=False)

# Create a line plot for Average Confidence by Itemset Size
plt.figure(figsize=(10, 6))
plt.plot(avg_confidence['itemset_size'], avg_confidence['confidence'], marker='o', linestyle='-', color='b')

# Adding labels and title
plt.xlabel('Itemset Size')
plt.ylabel('Average Confidence')
plt.title('Average Confidence by Itemset Size (Descending)')

# Reverse the x-axis
plt.gca().invert_xaxis()

# Add grid for readability
plt.grid(True, linestyle='--', alpha=0.7)

# Show the plot
plt.tight_layout()
plt.show()
```

The graph indicates that as the itemset size decreases, the average confidence tends to increase.

This makes sense because smaller itemsets are likely to have higher confidence levels, as the likelihood of a smaller set of items being bought together is higher compared to larger itemsets.
